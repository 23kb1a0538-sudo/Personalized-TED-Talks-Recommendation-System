# Personalized TED Talks Recommendation System
### SpringBoard Internship Program Capstone Project

This notebook demonstrates the end-to-end Machine Learning pipeline:
1. **Data Ingestion & Preprocessing**: Merging 2,628 talks with transcripts, cleaning text, extracting 14 emotional dimensions.
2. **Exploratory Data Analysis (EDA)**: Engagement distributions, top tags, duration patterns.
3. **Feature Engineering & Vectorization**: TF-IDF, Latent Semantic Analysis (TruncatedSVD), and user-item sparse matrices.
4. **Recommendation Models**: Content-Based, SVD Collaborative Filtering, and Dynamic Hybrid blending.
5. **Evaluation**: Precision@K, Recall@K, NDCG@K, MAP@K, and alpha trade-off curves.
